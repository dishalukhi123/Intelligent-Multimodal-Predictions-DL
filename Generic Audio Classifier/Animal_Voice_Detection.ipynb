# Animal Voice Detection — Teachable Machine Audio Model

This notebook documents how to run the animal sound classifier trained with Google Teachable Machine. Your model classes are **Background Noise, Bird Chirping, Cat Meow, Cow Moo, and Dog Bark**.

The actual audio inference is performed by the exported TensorFlow.js model in the accompanying `index.html` app. This notebook helps you set up and launch that app.

## 1. Project files

The ready-to-run ZIP should contain:

- `index.html` — microphone-based audio prediction app
- `my_model/model.json` — model architecture
- `my_model/metadata.json` — class labels and metadata
- `my_model/weights.bin` — trained model weights

Keep these files in the same folder structure after extracting the ZIP.

## 2. Confirm the model metadata

The metadata file supplied with this project lists these labels:

1. Background Noise
2. Bird Chirping
3. Cat Meow
4. Cow Moo
5. Dog Bark

```python
import json
from pathlib import Path

metadata_path = Path('my_model/metadata.json')
if metadata_path.exists():
    metadata = json.loads(metadata_path.read_text(encoding='utf-8'))
    print('Model name:', metadata.get('modelName', 'Not specified'))
    print('Class labels:')
    for label in metadata.get('wordLabels', []):
        print('-', label)
else:
    print('metadata.json not found. Extract the project ZIP and run this notebook from the project folder.')
```

## 3. Start the web app

Run the following cell from the folder containing `index.html`. Then open **http://localhost:8000** in Chrome and click **Start**. Allow microphone permission when prompted.

If the server command reports that port 8000 is already in use, stop the other server or use a different port such as 8001.

```python
import subprocess
import sys
from pathlib import Path

if not Path('index.html').exists():
    print('index.html not found. Extract Animal_Voice_Detection_Ready_To_Run.zip and open this notebook from the extracted project folder.')
else:
    print('Starting local web server at http://localhost:8000')
    print('Stop the server with Ctrl+C in the terminal where it is running.')
    subprocess.run([sys.executable, '-m', 'http.server', '8000'])
```

## 4. Test the classifier

- Try one animal sound at a time.
- Test background noise and quiet conditions too.
- Compare confidence scores across all five classes.
- Results are predictions and may be incorrect if the sound is noisy or unlike the training examples.
