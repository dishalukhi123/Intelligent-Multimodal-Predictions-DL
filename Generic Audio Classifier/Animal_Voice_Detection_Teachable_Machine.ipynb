# Animal Voice Detection — Teachable Machine Audio

This notebook guides you through training an audio classifier in **Google Teachable Machine** and using the exported model in a small web app.

**Important:** Teachable Machine does the training in your browser. First train and export the model, then paste your model URL into `index.html`.

## 1. Train the audio model

1. Open [Teachable Machine](https://teachablemachine.withgoogle.com/).
2. Choose **Get Started → Audio Project** (or the sound project option shown on the site).
3. Create classes such as:
   - `Dog Bark`
   - `Cat Meow`
   - `Bird Chirping`
   - `Cow Moo`
   - `Other / Background`
4. Record several examples for every class, in different environments. Include silence/background sounds.
5. Click **Train Model** and test it with new sounds.
6. Choose **Export Model → TensorFlow.js → Upload my model** (wording may vary), then copy the model URL.

## 2. Configure the web app

Open `index.html` from this project folder and replace `PASTE_YOUR_TEACHABLE_MACHINE_MODEL_URL_HERE/` with your exported model URL. Keep the trailing slash. The app expects the URL to provide `model.json` and `metadata.json`.

## 3. Run the app

The simplest option is to open `index.html` in Chrome after adding the model URL. If the browser blocks microphone access from a local file, run a local server from this folder:

```bash
python -m http.server 8000
```

Then open `http://localhost:8000` in Chrome and allow microphone access.

## 4. Tips for better animal sound detection

- Record multiple examples per animal and avoid recording only one very similar sound.
- Keep class examples balanced in length and quantity.
- Add a background/no-animal class to reduce false detections.
- Test in a different room and with sounds not used during training.
- The prediction is a model estimate, not a guarantee of the animal's identity.
