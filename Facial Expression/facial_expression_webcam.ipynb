# Facial Expression Recognition – Teachable Machine

Real-time webcam prediction using `keras_model.h5` and `labels.txt`.

Classes: Angry, Disgust, Fear, Happy, Neutral, Sad, Surprised.

````python
import os
import cv2
import numpy as np
import tf_keras

np.set_printoptions(suppress=True)

MODEL_PATH = os.path.join(os.getcwd(), "keras_model.h5")
LABELS_PATH = os.path.join(os.getcwd(), "labels.txt")

print("Model exists:", os.path.exists(MODEL_PATH))
print("Labels exist:", os.path.exists(LABELS_PATH))
````

````python
model = tf_keras.models.load_model(MODEL_PATH, compile=False)

with open(LABELS_PATH, "r", encoding="utf-8") as f:
    class_names = [line.strip() for line in f if line.strip()]

print("Model loaded successfully.")
print("Classes:", class_names)
````

````python
# Press ESC to stop the webcam
camera = cv2.VideoCapture(0)

if not camera.isOpened():
    print("❌ Camera could not be opened. Try VideoCapture(1).")
else:
    print("✅ Camera started — press ESC to stop")
    while True:
        ret, frame = camera.read()
        if not ret:
            print("❌ Could not read frame from camera.")
            break

        image = cv2.resize(frame, (224, 224), interpolation=cv2.INTER_AREA)
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        image = np.asarray(image, dtype=np.float32).reshape(1, 224, 224, 3)
        image = (image / 127.5) - 1.0

        prediction = model.predict(image, verbose=0)[0]
        index = int(np.argmax(prediction))
        confidence = float(prediction[index]) * 100

        class_name = class_names[index]
        if len(class_name) > 2 and class_name[0].isdigit() and class_name[1].isspace():
            class_name = class_name[2:].strip()

        text = f"{class_name} - {confidence:.1f}%"
        cv2.putText(frame, text, (20, 45), cv2.FONT_HERSHEY_SIMPLEX, 1, (0,255,0), 2, cv2.LINE_AA)
        cv2.imshow("Facial Expression Recognition", frame)

        if cv2.waitKey(1) & 0xFF == 27:
            break

    camera.release()
    cv2.destroyAllWindows()
    print("Camera stopped.")
````
